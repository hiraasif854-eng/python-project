```python
# Task 11: Personal Expense Tracker
total_expenses = int(input("Enter total expenses: "))
expenses = []
for i in range(total_expenses):
  name = input("Enter expense name: ")
  category = input("Enter expense category: ")
  amount = float(input("Enter amount: "))

  expense = {
      "name" : name,
      "category" : category,
      "amount" : amount
  }

  expenses.append(expense)

total_spending = 0
for expense in expenses:
  total_spending += expense["amount"]
category_spending = {}
for expense in expenses:
  if expense["category"] in category_spending:
    category_spending[expense["category"]] += expense["amount"]
  else:
    category_spending[expense["category"]] = expense["amount"]
highest_expense = expenses[0]
for expense in expenses:
  if expense["amount"] > highest_expense["amount"]:
    highest_expense = expense
budget = float(input("Enter your budget: "))
if total_spending > budget:
    print("Warning: Budget exceeded!")
else:
  print("you are with in your budget")
def display_report():
    print("Total Spending:", total_spending)
    print("Category Spending:", category_spending)
    print("Highest Expense:", highest_expense)
    print("Budget:", budget)
display_report()
```

```python
# Task 12: Store Inventory Manager

inventory = {}


def add_product():
    name = input("Enter product name: ")
    price = float(input("Enter product price: "))
    quantity = int(input("Enter quantity: "))

    if name in inventory:
        print("Product already exists")
    else:
        inventory[name] = {
            "price": price,
            "quantity": quantity
        }
        print("Product added successfully")


def sell_product():
    name = input("Enter product name: ")
    quantity = int(input("Enter quantity to sell: "))

    if name in inventory:
        if inventory[name]["quantity"] >= quantity:
            inventory[name]["quantity"] -= quantity
            print("Product sold successfully")
        else:
            print("Insufficient stock")
    else:
        print("Product not found")


def restock_product():
    name = input("Enter product name: ")
    quantity = int(input("Enter quantity to restock: "))

    if name in inventory:
        inventory[name]["quantity"] += quantity
        print("Product restocked successfully")
    else:
        print("Product not found")


def search_product():
    name = input("Enter product name: ")

    if name in inventory:
        print("Price:", inventory[name]["price"])
        print("Quantity:", inventory[name]["quantity"])
    else:
        print("Product not found")


def display_inventory():
    print("Inventory:", inventory)

    print("Low-stock products:")
    for name, product in inventory.items():
        if product["quantity"] < 5:
            print(name, product)

    total_value = 0
    for name, product in inventory.items():
        total_value += product["price"] * product["quantity"]

    print("Total Inventory Value:", total_value)


while True:
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        break

    else:
        print("Invalid choice")
```

```python
# Task 13:
# Task 13: Library Book Management System

books = []


def add_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists")
            return

    title = input("Enter book title: ")
    author = input("Enter author: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "status": "Available"
    }

    books.append(book)
    print("Book added successfully")


def display_books():
    for book in books:
        print(book)


def search_book():
    title = input("Enter book title: ")

    for book in books:
        if book["title"] == title:
            print(book)
            return

    print("Book not found")


def borrow_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            if book["status"] == "Available":
                book["status"] = "Borrowed"
                print("Book borrowed successfully")
            else:
                print("Book is already borrowed")
            return

    print("Book not found")


def return_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            book["status"] = "Available"
            print("Book returned successfully")
            return

    print("Book not found")


while True:
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        break

    else:
        print("Invalid choice")
```

```python
# Task 14: Restaurant Ordering System

menu= {
    "Burger":500,
    "Pizza":1200,
    "Pasta":800,
    "Drink":200
}

orders = []

print("========= Restaurant Menu =========")
for key,value in menu.items():
  print(f'{key} : {value}')

while True:
  item_name = input("Enter your item name: ").title()
  if item_name in menu:
    quantity = int(input("Enter quantity of your item: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })
  else:
    print("This item is not available")

  choice = input("You want to add more items: yes/no: ")
  if choice == "no":
    break

total = 0
for order in orders:
  total = total + order["sub_total"]

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1
else:
  discount = 0

grand_total = total + tax - discount

def receipt():
  print("\n============ Receipt ============")

  for order in orders:
    print(f'{order["item"]} x {order["quantity"]} = {order["sub_total"]}')

  print("Total: ", total)
  print("Tax: ", tax)
  print("Discount: ", discount)
  print("Grand Total: ", grand_total)

receipt()
```

```python
# Task 15: Clinic Appointment Manager
patients = []

def add_patient():
  patient_id = int(input("Enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient already exists")
      return
  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is emergency pateint? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency
  })

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient found")
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')
      print(f'Patient emergency: {patient["emergency"]}')
      return
    else:
      print("Patient not found")

def display_appointment():
  print("========== Emergency pateint ==========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("========== normal pateint ==========")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

def delete_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      patients.remove(patient)
      print("Patient deleted")

def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1

while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Delete patient")
  print("5. Count patient")
  print("6. Exit")

  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    delete_patient()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    break
  else:
    print("Invalid choice")
```

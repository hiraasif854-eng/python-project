```python
# Concat:
df1 = {
    "Customer Name" : ["Ali", "Sara", "Ahmed", "Zaid"],
    "City" : ["Karachi", "Islamabad", "Karachi", "Lahore"]
}
df2 = {
    "Customer Name" : ["Ahad", "Saira", "Ayesha", "Zahid"],
    "City" : ["Karachi", "Islamabad", "Multan", "Lahore"]
}
df1 = pd.DataFrame(df1)
df2 = pd.DataFrame(df2)
final_result = pd.concat([df1, df2], ignore_index=True)
print(final_result)
```

```python
# Merge:
Employees = pd.DataFrame({
    "ID": [1, 2, 3],
    "Employee": ["Ayan", "Sana", "Aisha"]
})

Salary = pd.DataFrame({
    "ID": [1, 2, 3],
    "Salary": [8000, 9000, 7000]
})
final_df = pd.merge(Employees, Salary, on="ID")

print(final_df)
```

```python
# important array properties
#1) ndim
import numpy as np
data = np.array([
    [10,20,30],
    [40,50,60]
])
print(data.ndim)
```

```python
# 2) shape
# shape tells us the number of rows and columns
import numpy as np
data = np.array([
    [10,20,30],
    [40,50,60]
])
print(data.shape)
```

```python
# 3) size
# size tells us the total number of values
data = np.array([
    [10,20,30],
    [40,50,60]
])
print(data.size)
```

```python
# 4) dtype:
# dtype tells us the data type
import numpy as np
data = np.array([
    [10,20,30],
    [40,50,60]
])
print(data.dtype)
```

```python
# 5) Indexing
# indexing means getting one specific value from an array
# indexing starts from 0
import numpy as np
numbers = np.array([10,20,30,40,50])
print(numbers[0])
print(numbers[3])
print(numbers[2])
print(numbers[1])
print(numbers[4])
# negative index
print(numbers[-1])
```

```python
# 6) indexing in a 2D array
marks = np.array([
    [70,80,90],
    [60,75,85]
])
print(marks[0 , 1])
print(marks[1,2])
```

```python
# 7) Slicing
# slicing means getting multiple values from an array
numbers = np.array([10,20,30,40,50])
print(numbers[1:4]) # the pattern is [start : stop]
print(numbers[1:3]) # the start value is included, stop value not included
```

```python
# 8) Vectorized operations
# important features of numpy
# NumPy can perform one operation on all values at the same time
numbers = np.array([10,20,30,40,50])
# ADD
print(numbers + 5)
# MULTIPLY
print(numbers * 5)
# SUBTRACT
print(numbers - 5)
# DIVIDE
print(numbers / 2)
# This is called a vectorized operation.
# We do not need to use a loop for every value.
```

```python
# 9) Operations between two arrays
# We can also perform operations between two arrays
a = np.array([10,20,30])
b = np.array([1,2,3])
# ADDITION
print(a + b)
# MULTIPLICATION
print(a * b)
# this is called as an element wise operation
```

```python
# 10) Mathematical functions:
# Square Root:
numbers = np.array([4,9,16,25])
print(np.sqrt(numbers))
# Power:
numbers = np.array([2,3,4])
print(np.power(numbers , 2))
# Absolute Value:
numbers = np.array([-10,-5,4,8]) # gives positive value
print(np.abs(numbers))
# Round:
numbers = np.array([4.2 , 5.7 , 8.9])
print(np.round(numbers))
```

```python
# 11) Aggregration Functions
# Aggregration function gives us summary values
marks = np.array([70,85,60,95,80])
# SUM
print(np.sum(marks))
# AVERAGE:
print(np.mean(marks))
# MAXIMUM VALUE:
print(np.max(marks))
# MINIMUM VALUE:
print(np.min(marks))
# MEDIAN:
print(np.median(marks)) # first arrange in asecnding to descending order,
# then finding middle value
# STANDARD DEVIATION: tells us how spread out the values are
print(np.std(marks))
```

```python
# 12) Conditions in NumPy:
# We can apply conditions directly to NumPy arrays.
marks = np.array([30, 50, 80, 45, 90, 20])
print(marks >= 50)
```

```python
# 13) Filtering Data:
# We can also get only the values that match a condition.
marks = np.array([30, 50, 80, 45, 90, 20])
passed = marks[marks >= 50]
print(passed)
# To get failed marks:
failed = marks[marks < 50]
print(failed)


```

```python
# 14) Reshape:
# reshape() changes the structure of an array
numbers = np.array([1, 2, 3, 4, 5, 6])
new_array = numbers.reshape(2,3)
print(new_array)
#Important rule:
# The total number of values must stay the same.
# 2 × 3 = 6
# So the reshape is possible.
```

```python
# 15. Flatten
# flatten() changes a multi-dimensional array into a one-dimensional array.
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(data.flatten())
```

```python
# 16. Useful NumPy Functions
# zeros()
# Creates an array with zero values
print(np.zeros(5))
# ones()
# Creates an array with one values.
print(np.ones(5))
# arange()
#Creates a range of numbers.
print(np.arange(1,11))
# using a step:
print(np.arange(0,11,2))# [start, stop, step]
```
